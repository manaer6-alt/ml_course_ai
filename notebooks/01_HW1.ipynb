6. Is it true that people who earn more than 50K have at least high school education? (education – Bachelors, Prof-school, Assoc-acdm, Assoc-voc, Masters or Doctorate feature)

```python
df['education'].value_counts()

below_HS = [
    "Preschool",
    "1st-4th",
    "5th-6th",
    "7th-8th",
    "9th",
    "10th",
    "11th",
    "12th",
]

def is_high_salary_without_HS(df):
    if ((df['salary'] == '>=50K') & (~df['education'].isin(below_HS))).any():
        print("Yes")
    else:
        print("No")

is_high_salary_without_HS(df)
  
```

8. Among whom is the proportion of those who earn a lot (>50K) greater: married or single men (marital-status feature)? Consider as married those who have a marital-status starting with Married (Married-civ-spouse, Married-spouse-absent or Married-AF-spouse), the rest are considered bachelors.

```python
df['marital-status'].value_counts()

df['married_or_single'] = df['marital-status'].apply(lambda x: 'Married' if x.startswith('Married') else 'Single')

df[(df['sex'] == 'Male') & (df['salary'] == '>50K')]['married_or_single'].value_counts()

df_find_men = df[(df['sex'] == 'Male') & (df['salary'] == '>50K')]['married_or_single'].value_counts()

df_find_men
```

```python
if df_find_men['Married'] > df_find_men['Single']:
    print("Married")
else:
    print("Single")
```

9. What is the maximum number of hours a person works per week (hours-per-week feature)? How many people work such a number of hours, and what is the percentage of those who earn a lot (>50K) among them?

```python
df['hours-per-week'].value_counts().sort_index(ascending=False)

df_hours = df['hours-per-week'].value_counts().sort_index(ascending=False)

df_hours.index.max()
```

```python
df_hours.loc[df_hours.index.max()]
```

```python
df[df['hours-per-week'] == df_hours.index.max()]['salary'].value_counts(normalize=True)['>50K']
```
